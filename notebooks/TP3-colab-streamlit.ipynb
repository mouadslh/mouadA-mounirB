## 5. Lancer Streamlit en tâche de fond

```python
import subprocess
import threading

def lancer_streamlit():
    subprocess.run(["streamlit", "run", "ui/app.py", "--server.port", "8501"])

# Lancer Streamlit dans un thread
thread = threading.Thread(target=lancer_streamlit, daemon=True)
thread.start()

print("Streamlit démarré sur le port 8501")
```
